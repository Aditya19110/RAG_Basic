### TXT Loader

```python
### TextLoader

from langchain_community.document_loaders import TextLoader
loader = TextLoader("../data/text_files/python_intro.txt", encoding = "utf-8")
document=loader.load()
print(document)
```

### DIR Loader TXT

```python
### Dir Loader

from langchain_community.document_loaders import DirectoryLoader
dir_loader = DirectoryLoader(
    "../data/text_files",
    glob = "**/*.txt", #pattern matching
    loader_cls = TextLoader, ##loader class to be used
    loader_kwargs = {'encoding':'utf-8'}
)
document = dir_loader.load()
```

### PDF Dir Loader

```python
from langchain_community.document_loaders import PyPDFLoader, PyMuPDFLoader
from langchain_community.document_loaders import DirectoryLoader
dir_loader = DirectoryLoader(
    "../data/pdf_files",
    glob = "**/*.pdf", #pattern matching
    loader_cls = PyMuPDFLoader, ##loader class to be used
)
pdf_docs = dir_loader.load()
pdf_docs
```

### DIR Loader CSV

```python
from langchain_community.document_loaders import CSVLoader,DirectoryLoader
dir_loader = DirectoryLoader(
    "../data/csv_files",
    glob = "**/*.csv", #pattern matching
    loader_cls = CSVLoader, ##loader class to be used
)
csv_docs = dir_loader.load()
csv_docs
```

### DIr Loader Excel

```python
from langchain_community.document_loaders import UnstructuredExcelLoader,DirectoryLoader
dir_loader = DirectoryLoader(
    "../data/excel_files",
    glob = "**/*.xlsx", #pattern matching
    loader_cls = UnstructuredExcelLoader, ##loader class to be used
)
xlsx_docs = dir_loader.load()
xlsx_docs
```
